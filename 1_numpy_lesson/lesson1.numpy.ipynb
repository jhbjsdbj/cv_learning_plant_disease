```python
# 第一步：把我们之前装好的numpy和matplotlib工具从仓库里拿出来（就像做饭前先把案板、盘子从柜子里掏出来）
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams["font.family"]=["SimHei"]
plt.rcParams["axes.unicode_minus"]=False
# 第二步：我们手动做一张迷你“测试图片”——就4个像素点，小到你一眼能数清楚
# 每个方括号里的三个数，对应一个像素的红、绿、蓝三个颜色的亮度，0是最暗，255是最亮
mini_image = np.array([
    [[255, 0, 0],   [0, 255, 0]],   # 第一行像素：左边纯红，右边纯绿
    [[0, 0, 255],   [255, 255, 255]]# 第二行像素：左边纯蓝，右边纯白
])

# 第三步：打印这张“图片”在计算机眼里长啥样——就是一堆数字
print("计算机眼里的图片数组：")
print(mini_image)

# 第四步：打印这张图片的形状——也就是它的尺寸
print("这张图的尺寸shape是：", mini_image.shape)

# 第五步：把这堆数字，重新变成人眼能看懂的图片显示出来
plt.imshow(mini_image)
plt.axis('off') # 把坐标轴去掉，只看图片
plt.show()
```

```python
print("原始的四个像素点：")
print(mini_image)
plt.imshow(mini_image)
plt.axis('off')
plt.show()
```

```python
# 块2：挑第一行第一个像素——就是左上角的纯红块
pixel_red = mini_image[0, 0]
print("挑出来的红像素：", pixel_red)
# 块3：挑第一行所有像素——就是上面那排的红+绿
row_top = mini_image[0, :]
print("挑出来的第一行：", row_top)
# 块4：挑第二列所有像素——就是右边那列的绿+白
col_right = mini_image[:, 1]
print("挑出来的第二列：", col_right)
```

```python
# 块5：裁左上角1x1的红块
crop_red = mini_image[0:1, 0:1]
print("裁出来的红块shape是：", crop_red.shape)
plt.imshow(crop_red)
plt.axis('off')
plt.show()
```

```python
plt.imshow(mini_image)
plt.axis('off')
plt.show()
```

```python
pixel_blue = mini_image[1:2,0:1]
print("蓝色像素点的尺寸shape为:",pixel_blue.shape)
pixel_white = mini_image[1:2,1:2]
plt.imshow(pixel_white)
plt.axis('off')
plt.show
```

```python
print("接下来进行亮度提升：")
plt.imshow(mini_image)
plt.axis('off')
plt.title("原图")
plt.show()
bright_image =np.clip(mini_image + 50 , a_min=0 , a_max=255) 

plt.Figure(figsize=(6,3))
plt.subplot(1,2,1)
plt.imshow(mini_image)
plt.title("原图")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(bright_image)
plt.title("亮度提高后")
plt.axis('off')
plt.show()
```

```python
print("只对左上角像素进行处理：")
change_red_pixel=mini_image[0:1,0:1]
change_red_pixel[0,0,2]=np.clip(change_red_pixel[0,0,2]+150 , a_min=0 , a_max=255)
plt.imshow(change_red_pixel)
plt.axis('off')
plt.show()
```
